```python
#Importando
import nltk
import spacy
from nltk.tokenize import word_tokenize
#from nltk.corpus import stopwords
from nltk.util import ngrams
#from nltk.probability import FreqDist
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nlp = spacy.load("pt_core_news_sm")
```

## Tokenização Spacy

```python
# Função para tokenizar texto usando spaCy e retornar a lista de tokens
def spacy_tokenize_list(text):
    doc = nlp(text)
    return [token.text for token in doc]

# Aplicar o tokenizador spaCy e armazenar a lista de tokens
df['Tokens_spacy'] = df['texto_integrado_original'].apply(spacy_tokenize_list)

# Contar o número de tokens gerados pelo spaCy
df['NTokens_spacy'] = df['Tokens_spacy'].apply(len)

# Exibir as primeiras linhas do DataFrame com as novas colunas
display(df[['texto_integrado_original', 'Tokens_spacy', 'NTokens_spacy']].head())
```

### Contagem de Palavras (excluindo pontuação e espaços) com spaCy

Vamos definir uma função para contar o número de palavras em um texto, utilizando o modelo spaCy para filtrar tokens que não são pontuação ou espaços. Em seguida, aplicaremos essa função às colunas `texto_integrado_original`.

```python
# Função para contar palavras usando spaCy, excluindo pontuação e espaços
def count_words_spacy(text):
    doc = nlp(text)
    words = [
        token.text for token in doc
        if not token.is_punct and not token.is_space
    ]
    return len(words)

def count_unique_words_spacy(text):
    doc = nlp(text)
    unique_words = set(
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
    )
    return len(unique_words)

def diversidade_lexical(text):
  count_words_spacy(text)
  count_unique_words_spacy(text)
  return count_unique_words_spacy(text)/count_words_spacy(text)

# Aplicar a função à coluna 'texto_integrado_original'
df['NWords_original'] = df['texto_integrado_original'].apply(count_words_spacy)

# Aplicar a função à coluna 'texto_integrado_original'
df['NWords_unica_original'] = df['texto_integrado_original'].apply(count_unique_words_spacy)

# Diversidade Lexical
df['Diversidade_lexical'] = df['texto_integrado_original'].apply(diversidade_lexical)

# Exibir as primeiras linhas do DataFrame com as novas colunas de contagem de palavras
display(df[['texto_integrado_original', 'NTokens_spacy','NWords_original','NWords_unica_original','Diversidade_lexical']].head())
```

##Contagem de Classes gramáticais
### POS-tagging

```python

def analisar_classes_gramaticais(text):
    doc = nlp(text)
    # Inicializa um dicionário para armazenar contagens E listas de palavras
    results = {
        'Verbos': 0, 'Verbos_palavras': [],
        'Substantivos': 0, 'Substantivos_palavras': [],
        'Adjetivos': 0, 'Adjetivos_palavras': [],
        'Advérbios': 0, 'Advérbios_palavras': [],
        'Pronomes': 0, 'Pronomes_palavras': [],
        'Preposicoes': 0, 'Preposicoes_palavras': []
    }

    # Percorre os tokens e conta/armazena as classes gramaticais
    for token in doc:
        if token.pos_ == "VERB":
            results['Verbos'] += 1
            results['Verbos_palavras'].append(token.text)
        elif token.pos_ == "NOUN":
            results['Substantivos'] += 1
            results['Substantivos_palavras'].append(token.text)
        elif token.pos_ == "ADJ":
            results['Adjetivos'] += 1
            results['Adjetivos_palavras'].append(token.text)
        elif token.pos_ == "ADV":
            results['Advérbios'] += 1
            results['Advérbios_palavras'].append(token.text)
        elif token.pos_ == "PRON":
            results['Pronomes'] += 1
            results['Pronomes_palavras'].append(token.text)
        elif token.pos_ == "ADP": # ADP é a tag universal para Adposition (Preposições e Pós-posições)
            results['Preposicoes'] += 1
            results['Preposicoes_palavras'].append(token.text)

    return results


# Aplicar a função e expandir o dicionário em novas colunas
grammatical_analysis_results = df['texto_integrado_original'].apply(analisar_classes_gramaticais)

# Converter a série de dicionários em um DataFrame
expanded_grammatical_df_new = pd.json_normalize(grammatical_analysis_results)

# Renomear as colunas para o padrão desejado: 'Analise_classes_Classe' para contagens e 'Analise_classes_Classe_palavras' para as listas
new_columns_map = {}
for col in expanded_grammatical_df_new.columns:
    if '_palavras' in col:
        new_columns_map[col] = 'Analise_classes_' + col
    else:
        # Para as colunas de contagem (e.g., 'Verbos'), usamos o nome 'Analise_classes_Verbos'
        new_columns_map[col] = 'Analise_classes_' + col

expanded_grammatical_df_new = expanded_grammatical_df_new.rename(columns=new_columns_map)

# Remover colunas de análise gramatical existentes para evitar duplicatas antes de concatenar as novas
existing_analysis_cols = [col for col in df.columns if col.startswith('Analise_classes_')]
df = df.drop(columns=existing_analysis_cols, errors='ignore') # 'errors=ignore' evita erro se a coluna não existir

# Concatenar as novas colunas ao DataFrame original
df = pd.concat([df, expanded_grammatical_df_new], axis=1)

# Exibir as primeiras linhas do DataFrame com as novas colunas
display_cols = [
    'texto_integrado_original',
    'Analise_classes_Verbos', 'Analise_classes_Verbos_palavras',
    'Analise_classes_Substantivos', 'Analise_classes_Substantivos_palavras',
    'Analise_classes_Adjetivos', 'Analise_classes_Adjetivos_palavras',
    'Analise_classes_Advérbios', 'Analise_classes_Advérbios_palavras',
    'Analise_classes_Pronomes', 'Analise_classes_Pronomes_palavras',
    'Analise_classes_Preposicoes', 'Analise_classes_Preposicoes_palavras'
]
display_cols_final = [col for col in display_cols if col in df.columns] # Filtra apenas colunas que realmente existem
display(df[display_cols_final].head())
```

```python
## 8º PRONOMES DE DIÁLOGO
# dialog_pronoun_ratio

def dialog_pronoun(text):
    doc = nlp(text)
    # Avoid division by zero if the text is empty
    if len(doc) == 0:
        return 0.0

    dialog_pronouns = [
        token for token in doc
        if token.pos_ == "PRON"
        and token.morph.get("Person") in [["1"], ["2"], ["3"]]
        ]
    return dialog_pronouns

    # Calculate the ratio
def dialog_pronoun_ratio(text):
    doc = nlp(text)
    dialog_pronouns = dialog_pronoun(text)
    dialog_pronoun_ratio = len(dialog_pronouns) / len(doc)
    return dialog_pronoun_ratio

# Apply the function to the 'texto_integrado_original' column
df['dialog_pronouns'] = df['texto_integrado_original'].apply(dialog_pronoun)
df['dialog_pronoun_ratio'] = df['texto_integrado_original'].apply(dialog_pronoun_ratio)
# Display the first lines of the DataFrame with the new column
# The original display call tried to access 'dialog_pronouns' as a column, which was not created.
display(df[['texto_integrado_original', 'dialog_pronouns', 'dialog_pronoun_ratio']].head(25))
```
