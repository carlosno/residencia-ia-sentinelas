```python
# ==============================================================================
# PROJETO SENTINELA — ESTUDO DE ABLAÇÃO DE FEATURES (5-FOLD CV)
# Alvos: 1. golpe_original (Binário) | 2. classe_curada (Multiclasse - 3 classes)
# Modelos: Linear SVM, Random Forest e Complement Naive Bayes
# Base: Exclusivamente 'treino_balanceado_1_1_1' (168 amostras balanceadas)
# ==============================================================================

# 1. INSTALAÇÃO E IMPORTAÇÃO DE PACOTES
!pip -q install pandas numpy openpyxl spacy scikit-learn matplotlib seaborn
!pip install -q https://github.com/explosion/spacy-models/releases/download/pt_core_news_sm-3.8.0/pt_core_news_sm-3.8.0-py3-none-any.whl

import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from matplotlib.patches import Patch

try:
    import pt_core_news_sm
    nlp = pt_core_news_sm.load()
except Exception:
    import spacy
    nlp = spacy.load("pt_core_news_sm")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import ComplementNB
from sklearn.model_selection import StratifiedKFold, cross_validate

warnings.filterwarnings("ignore")
RANDOM_STATE = 42
print("Ambiente configurado e spaCy carregado com sucesso.")

# ==============================================================================
# 2. CARREGAMENTO EXCLUSIVO DA ABA TREINO BALANCEADO (168 AMOSTRAS)
# ==============================================================================
base_url = (
    "https://github.com/carlosno/residencia-ia-sentinelas/"
    "blob/main/data/processed/base_golpes_sentinela_balanceadateste.xlsx?raw=true"
)

df = pd.read_excel(base_url, sheet_name="treino_balanceado_1_1_1")
coluna_texto = "texto_integrado_original"

print(f"\nBase carregada: {df.shape[0]} instâncias.")
print(f"Distribuição do Alvo 1 (golpe_original):\n{df['golpe_original'].value_counts()}")
print(f"\nDistribuição do Alvo 2 (classe_curada):\n{df['classe_curada'].value_counts()}")

# ==============================================================================
# 3. EXTRAÇÃO DAS 15 FEATURES CANDIDATAS
# ==============================================================================
URL_PATTERN = re.compile(r"(https?://|www\.)[^\s]+", re.IGNORECASE)
EMAIL_PATTERN = re.compile(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b")
TELEFONE_PATTERN = re.compile(r"(?:\+55\s?)?(?:\(?\d{2}\)?[\s.-]?)?9?\d{4}[\s.-]?\d{4}")
TERMOS_URGENCIA = [
    "urgente", "imediato", "imediatamente", "agora", "última chance", "bloqueado",
    "bloqueada", "suspenso", "suspensa", "prazo", "atenção", "atencao", "pix"
]

print("\nExtraindo metadados, comportamentais e métricas estilométricas do spaCy...")

# Metadados
df["tem_url"] = df[coluna_texto].apply(lambda t: int(bool(URL_PATTERN.search(str(t)))))
df["tem_email"] = df[coluna_texto].apply(lambda t: int(bool(EMAIL_PATTERN.search(str(t)))))
df["tem_telefone"] = df[coluna_texto].apply(lambda t: int(bool(TELEFONE_PATTERN.search(str(t)))))

# Comportamentais
df["tem_urgencia"] = df[coluna_texto].apply(lambda t: int(any(u in str(t).lower() for u in TERMOS_URGENCIA)))
df["contagem_digitos"] = df[coluna_texto].apply(lambda t: sum(c.isdigit() for c in str(t)))
df["contagem_exclamacoes"] = df[coluna_texto].apply(lambda t: str(t).count("!"))
df["razao_maiusculas"] = df[coluna_texto].apply(
    lambda t: (sum(c.isupper() for c in str(t) if c.isalpha()) / len([c for c in str(t) if c.isalpha()]))
    if any(c.isalpha() for c in str(t)) else 0.0
)

# Estilometria spaCy
dens_verbos, dens_subst, dens_adj, dens_pron = [], [], [], []
razao_imp, razao_pron2a, tem_imp_raiz, div_lex = [], [], [], []

for doc in nlp.pipe(df[coluna_texto].fillna("").astype(str), batch_size=50):
    words = [t for t in doc if not t.is_punct and not t.is_space]
    n_w = len(words)

    if n_w == 0:
        dens_verbos.append(0.0); dens_subst.append(0.0); dens_adj.append(0.0); dens_pron.append(0.0)
        razao_imp.append(0.0); razao_pron2a.append(0.0); tem_imp_raiz.append(0); div_lex.append(0.0)
        continue

    nv = sum(1 for t in words if t.pos_ == "VERB")
    np_ = sum(1 for t in words if t.pos_ == "PRON")
    na = sum(1 for t in words if t.pos_ == "ADJ")
    ns = sum(1 for t in words if t.pos_ == "NOUN")

    v_imp = sum(1 for t in words if t.pos_ == "VERB" and "Imp" in t.morph.get("Mood"))
    p_2a = sum(1 for t in words if t.pos_ == "PRON" and ("2" in t.morph.get("Person") or t.text.lower() in ["você", "voce", "seu", "sua", "te"]))

    imp_raiz = int(any(
        t.dep_ == "ROOT" and (
            (t.pos_ == "VERB" and "Imp" in t.morph.get("Mood")) or
            t.lemma_ in ["clicar", "acessar", "atualizar", "regularizar", "baixar", "pagar", "confirmar"]
        ) for t in doc
    ))

    lemmas = set(t.lemma_.lower() for t in words if t.is_alpha)

    dens_verbos.append(nv / n_w)
    dens_subst.append(ns / n_w)
    dens_adj.append(na / n_w)
    dens_pron.append(np_ / n_w)
    razao_imp.append((v_imp / nv) if nv > 0 else 0.0)
    razao_pron2a.append((p_2a / np_) if np_ > 0 else 0.0)
    tem_imp_raiz.append(imp_raiz)
    div_lex.append(len(lemmas) / n_w)

df["densidade_verbos"] = dens_verbos
df["densidade_substantivos"] = dens_subst
df["densidade_adjetivos"] = dens_adj
df["densidade_pronomes"] = dens_pron
df["razao_imperativo"] = razao_imp
df["razao_pronomes_2a"] = razao_pron2a
df["tem_imperativo_raiz"] = tem_imp_raiz
df["Diversidade_lexical"] = div_lex

features_para_ablation = [
    "tem_url", "tem_email", "tem_telefone",
    "tem_urgencia", "contagem_digitos", "contagem_exclamacoes", "razao_maiusculas",
    "densidade_verbos", "densidade_substantivos", "densidade_adjetivos", "densidade_pronomes",
    "razao_imperativo", "razao_pronomes_2a", "tem_imperativo_raiz", "Diversidade_lexical"
]

for col in features_para_ablation:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

y_bin = df["golpe_original"].astype(int)
y_cat = df["classe_curada"].astype(str)

# ==============================================================================
# 4. MOTOR GENÉRICO DE ABLAÇÃO COM VALIDAÇÃO CRUZADA (5-FOLD)
# ==============================================================================
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

modelos_dict = {
    "Linear SVM": (LinearSVC(dual="auto", class_weight="balanced", random_state=RANDOM_STATE), "standard"),
    "Random Forest": (RandomForestClassifier(n_estimators=100, max_depth=10, class_weight="balanced", random_state=RANDOM_STATE), "standard"),
    "Complement Naive Bayes": (ComplementNB(), "minmax")
}

def rodar_estudo_ablatacao(target_series, tipo_alvo="binario"):
    tabelas_modelos = {}

    for nome_modelo, (estimador, scaler_tipo) in modelos_dict.items():
        # Baseline (Apenas TF-IDF)
        pipe_base = Pipeline([
            ("tfidf", TfidfVectorizer(max_features=5000, ngram_range=(1, 2), sublinear_tf=True)),
            ("clf", estimador)
        ])

        scorers = ["accuracy", "precision", "recall", "f1"] if tipo_alvo == "binario" else ["accuracy", "f1_macro", "f1_weighted"]
        scores_base = cross_validate(pipe_base, df[coluna_texto], target_series, cv=cv, scoring=scorers)

        acc_base = scores_base["test_accuracy"].mean()

        if tipo_alvo == "binario":
            f1_base = scores_base["test_f1"].mean()
            prec_base = scores_base["test_precision"].mean()
            rec_base = scores_base["test_recall"].mean()
            col_metrica_principal = "F1-Score"
            registro_base = {
                "Modelo": nome_modelo, "Feature": "[BASELINE] Apenas TF-IDF",
                "Acurácia": acc_base, "Precisão": prec_base, "Recall": rec_base,
                "F1-Score": f1_base, "Δ F1 vs Base": 0.0, "Veredito": "Referência"
            }
        else:
            f1_base = scores_base["test_f1_macro"].mean()
            f1_w_base = scores_base["test_f1_weighted"].mean()
            col_metrica_principal = "F1-Macro"
            registro_base = {
                "Modelo": nome_modelo, "Feature": "[BASELINE] Apenas TF-IDF",
                "Acurácia": acc_base, "F1-Macro": f1_base, "F1-Ponderado": f1_w_base,
                "Δ F1-Macro vs Base": 0.0, "Veredito": "Referência"
            }

        registros = [registro_base]

        # Teste One-by-One para cada feature
        for feat in features_para_ablation:
            scaler = MinMaxScaler() if scaler_tipo == "minmax" else StandardScaler()
            prep = ColumnTransformer([
                ("tfidf", TfidfVectorizer(max_features=5000, ngram_range=(1, 2), sublinear_tf=True), coluna_texto),
                ("num", scaler, [feat])
            ])
            pipe = Pipeline([("prep", prep), ("clf", estimador)])

            scores = cross_validate(pipe, df[[coluna_texto, feat]], target_series, cv=cv, scoring=scorers)
            acc = scores["test_accuracy"].mean()

            if tipo_alvo == "binario":
                f1 = scores["test_f1"].mean()
                prec = scores["test_precision"].mean()
                rec = scores["test_recall"].mean()
                delta = f1 - f1_base
                veredito = "MELHORA (+)" if delta > 0.005 else "PENALIZA (-)" if delta < -0.005 else "NEUTRA (=)"
                registros.append({
                    "Modelo": nome_modelo, "Feature": feat,
                    "Acurácia": acc, "Precisão": prec, "Recall": rec,
                    "F1-Score": f1, "Δ F1 vs Base": delta, "Veredito": veredito
                })
            else:
                f1_m = scores["test_f1_macro"].mean()
                f1_w = scores["test_f1_weighted"].mean()
                delta = f1_m - f1_base
                veredito = "MELHORA (+)" if delta > 0.005 else "PENALIZA (-)" if delta < -0.005 else "NEUTRA (=)"
                registros.append({
                    "Modelo": nome_modelo, "Feature": feat,
                    "Acurácia": acc, "F1-Macro": f1_m, "F1-Ponderado": f1_w,
                    "Δ F1-Macro vs Base": delta, "Veredito": veredito
                })

        df_res = pd.DataFrame(registros)
        sort_col = "Δ F1 vs Base" if tipo_alvo == "binario" else "Δ F1-Macro vs Base"
        tabelas_modelos[nome_modelo] = df_res.sort_values(by=sort_col, ascending=False).reset_index(drop=True)

    return tabelas_modelos

# ==============================================================================
# 5. EXECUÇÃO DO ESTUDO PARA OS DOIS ALVOS
# ==============================================================================
print("\n" + "=" * 95)
print("EXECUTANDO ESTUDO DE ABLAÇÃO: ALVO 1 — golpe_original (BINÁRIO)")
print("=" * 95)
ablation_bin = rodar_estudo_ablatacao(y_bin, tipo_alvo="binario")

for mod in ["Linear SVM", "Random Forest", "Complement Naive Bayes"]:
    print(f"\n--- TABELA DE ABLAÇÃO BINÁRIA: {mod} ---")
    display(ablation_bin[mod].round(4))

print("\n" + "=" * 95)
print("EXECUTANDO ESTUDO DE ABLAÇÃO: ALVO 2 — classe_curada (MULTICLASSE: 3 CLASSES)")
print("=" * 95)
ablation_cat = rodar_estudo_ablatacao(y_cat, tipo_alvo="multiclasse")

for mod in ["Linear SVM", "Random Forest", "Complement Naive Bayes"]:
    print(f"\n--- TABELA DE ABLAÇÃO CATEGÓRICA (classe_curada): {mod} ---")
    display(ablation_cat[mod].round(4))

# ==============================================================================
# 6. VISUALIZAÇÃO GRÁFICA COMPARATIVA: BINÁRIO vs MULTICLASSE
# ==============================================================================
def gerar_grafico_triplo(tabelas_dict, delta_col, titulo_geral):
    fig, axes = plt.subplots(1, 3, figsize=(24, 8), sharey=True)
    nomes = ["Linear SVM", "Random Forest", "Complement Naive Bayes"]

    for i, nome in enumerate(nomes):
        df_p = tabelas_dict[nome]
        df_p = df_p[df_p["Feature"] != "[BASELINE] Apenas TF-IDF"].sort_values(by="Feature").reset_index(drop=True)

        cores = [
            "#27ae60" if v == "MELHORA (+)" else "#e74c3c" if v == "PENALIZA (-)" else "#7f8c8d"
            for v in df_p["Veredito"]
        ]

        axes[i].barh(df_p["Feature"], df_p[delta_col], color=cores, edgecolor="black", alpha=0.85)
        axes[i].axvline(0, color="black", linestyle="--", linewidth=1.2)
        axes[i].set_title(f"{nome}\n{delta_col}", fontsize=12, fontweight="bold")
        axes[i].set_xlabel("Variação em Relação ao Baseline", fontsize=11)
        axes[i].grid(axis="x", alpha=0.3)

    legend_elements = [
        Patch(facecolor="#27ae60", edgecolor="black", label="Melhora (+0.005)"),
        Patch(facecolor="#7f8c8d", edgecolor="black", label="Neutra (|Δ| ≤ 0.005)"),
        Patch(facecolor="#e74c3c", edgecolor="black", label="Penaliza (-0.005)")
    ]
    fig.legend(handles=legend_elements, loc="upper center", bbox_to_anchor=(0.5, 0.98), ncol=3, frameon=True, fontsize=11)
    fig.suptitle(titulo_geral, fontsize=14, fontweight="bold", y=1.03)
    plt.tight_layout(rect=[0, 0, 1, 0.93])
    plt.show()

print("\nGerando gráficos de ablação...")
gerar_grafico_triplo(ablation_bin, "Δ F1 vs Base", "ABLAÇÃO NO ALVO 1: golpe_original (BINÁRIO)")
gerar_grafico_triplo(ablation_cat, "Δ F1-Macro vs Base", "ABLAÇÃO NO ALVO 2: classe_curada (MULTICLASSE)")

# ==============================================================================
# 7. MATRIZ CONSOLIDADA DE CONSENSO (AMBOS OS ALVOS E 3 MODELOS)
# ==============================================================================
consenso_rows = []
for feat in features_para_ablation:
    v_bin_svm = ablation_bin["Linear SVM"].loc[ablation_bin["Linear SVM"]["Feature"] == feat, "Veredito"].values[0]
    v_bin_rf  = ablation_bin["Random Forest"].loc[ablation_bin["Random Forest"]["Feature"] == feat, "Veredito"].values[0]
    v_bin_cnb = ablation_bin["Complement Naive Bayes"].loc[ablation_bin["Complement Naive Bayes"]["Feature"] == feat, "Veredito"].values[0]

    v_cat_svm = ablation_cat["Linear SVM"].loc[ablation_cat["Linear SVM"]["Feature"] == feat, "Veredito"].values[0]
    v_cat_rf  = ablation_cat["Random Forest"].loc[ablation_cat["Random Forest"]["Feature"] == feat, "Veredito"].values[0]
    v_cat_cnb = ablation_cat["Complement Naive Bayes"].loc[ablation_cat["Complement Naive Bayes"]["Feature"] == feat, "Veredito"].values[0]

    todos_vereditos = [v_bin_svm, v_bin_rf, v_bin_cnb, v_cat_svm, v_cat_rf, v_cat_cnb]

    if "MELHORA (+)" in todos_vereditos:
        recomendacao = "MANTER (Agrega Ganho)"
    elif todos_vereditos.count("PENALIZA (-)") >= 3:
        recomendacao = "DESCARTAR (Penaliza)"
    else:
        recomendacao = "NEUTRA (Opcional)"

    consenso_rows.append({
        "Feature": feat,
        "SVM (Bin)": v_bin_svm, "RF (Bin)": v_bin_rf, "CNB (Bin)": v_bin_cnb,
        "SVM (Cat)": v_cat_svm, "RF (Cat)": v_cat_rf, "CNB (Cat)": v_cat_cnb,
        "Decisão Final": recomendacao
    })

df_consenso_geral = pd.DataFrame(consenso_rows).sort_values(by="Decisão Final", ascending=False).reset_index(drop=True)

print("\n" + "=" * 105)
print("MATRIZ CONSOLIDADA DE DECISÃO (DOIS ALVOS E TRÊS MODELOS)")
print("=" * 105)
display(df_consenso_geral)
```
